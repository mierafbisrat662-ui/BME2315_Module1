## Disease Background


### Prevalence & Incidence

Alzheimer’s disease affects a substantial and growing portion of the older U.S. population. In 2026, an estimated 7.4 million Americans age 65 and older are living with clinical Alzheimer’s dementia, representing about 1 in 9 adults in this age group. The incidence of Alzheimer’s also increases significantly with age. Data from the Chicago Health and Aging Project estimated an annual incidence of approximately 0.4% among adults ages 65–74, rising to 3.2% among adults ages 75–84 and 7.6% among those age 85 and older. These trends demonstrate that age is strongly associated with the likelihood of developing Alzheimer’s disease and help explain why the overall number of Americans affected by the disease continues to increase as the population ages.

Source: (https://www.alz.org/alzheimers-dementia/facts-figures)

### Economic Burden

Alzheimer’s disease creates an economic burden for the individual and their family, the healthcare system and society. Costs include doctor visits, medications, hospitalizations, nursing facilities, and other expenses that build up over time. The lifetime cost of Alzheimer’s is approximately more than $405,000 per person.The total cost of health and long-term care for people with Alzheimer’s is projected to reach $409 billion in 2026. Medicare and Medicaid are expected to cover $263 billion of these costs, while individuals are expected to pay $103 billion out of pocket. Families have to provide unpaid, full time care and in order to do so they must reduce the hours that they work or quit their job altogether, which places them further into financial stress. Around 13 million unpaid caregivers provided 19.6 billion hours of care in 2025. 41% of caregivers have a household income of $50,000 or less and around 30% of caregivers are older than 65 years of age, which increases the difficulty of bringing in a stable income and also providing adequate care. 

Source: https://www.alz.org/alzheimers-dementia/facts-figures

### Risk Factors (Genetic, Lifestyle)

Several genetic, biological, and lifestyle factors can increase a person’s risk of developing Alzheimer’s disease. Older age is the greatest known risk factor, with most cases occurring after age 65 and risk increasing as a person gets older. Genetics also play an important role. Variants of the APOE gene can affect Alzheimer’s risk: the APOE ε4 allele increases risk, particularly when a person inherits two copies, while APOE ε2 may provide some protection. Having a family history of the disease is also associated with an increased risk, though neither APOE status nor family history determines whether someone will develop the disease.

Other factors associated with risk include traumatic brain injury, physical inactivity, smoking, an unhealthy diet, and inadequate sleep. Alzheimer’s is also more common among women: almost two-thirds of Americans living with Alzheimer’s are women, although some of this difference is explained by women living longer on average, and researchers continue to study whether biological differences contribute additional risk.

### Societal Determinants

Several social factors affect the risk of developing Alzheimer’s disease and access to care. Lower education attainment has been associated with poorer brain health. Socioeconomic status also plays a role in education opportunities, such as the ability to attend and pay for college. Less access to preventative healthcare and medical treatment can increase the risk of dementia. Regular healthcare can help support earlier diagnosis of Alzheimer’s and identify and manage other diseases that increase dementia risk (diabetes, cardiovascular disease). Housing, transportation, and environment can influence whether or not patients have access to healthy food, exercise opportunities, and other resources. Social isolation and loneliness can be associated with greater cognitive decline.

### Symptoms

Alzheimer’s disease causes progressive changes in memory, thinking, judgment, daily functioning, and behavior. Memory loss is one of the most common early symptoms, including forgetting recent events and conversations, repeating statements, misplacing belongings, and becoming lost even in familiar places.

As the disease progresses, people may also have difficulty concentrating and thinking, especially with abstract concepts or numbers, and may struggle with managing finances or completing multiple tasks at the same time. The disease can also impair judgment and decision-making, which can lead to poor choices or difficulty with everyday problems.

Changes in personality and behavior can also occur, including mood changes, depression, social withdrawal, irritability, anger, delusions, and changes in sleeping patterns.

Source: (https://www.mayoclinic.org/diseases-conditions/alzheimers-disease/symptoms-causes/syc-20350447)

### Diagnosis

Diagnosis has evolved from clinical assessments alone (memory loss, cognitive testing, and medical history) to a biomarker driven framework. Modern diagnostic procedures use neuroimaging, such as PET scans and fluid biomarkers to measure levels of tau accumulation and amyloid-beta plaques before brain damage occurs. MRI and CT scans can identify structural changes and help rule out other causes of cognitive impairment, including brain shrinkage, strokes, tumors, or fluid buildup. Cerebrospinal fluid obtained through a lumbar puncture can be tested for amyloid-beta 42, tau, and phosphorylated tau. Blood-based biomarkers are an important development because they provide a less invasive way to detect Alzheimer related changes. A combination of biomarker results, clinical symptoms, and medical history is used for diagnosis rather than independent consideration of each.

### Standard of Care Treatments & Reimbursement

Treatment for Alzheimer’s disease depends on the individual’s symptoms, disease severity, and overall health. Although there is currently no cure, treatment may include medications that temporarily improve memory and thinking, medications that manage behavioral symptoms, and newer therapies that target the underlying biology of the disease and may slow its progression in some patients.

As Alzheimer’s progresses, supportive care becomes increasingly important and may involve caregivers, simplifying daily routines, and modifying the home environment to remove hazards and improve safety.

In terms of reimbursement, Medicare generally covers medically necessary diagnostic testing, physician visits, prescription drugs under applicable plans, and certain skilled nursing or home-health services, but it does not generally pay for long-term custodial residential care. Medicaid may provide broader long-term-care coverage for eligible individuals, including nursing-home care and some home- and community-based services.

Source: (https://www.medicare.gov/basics/costs/help/medicaid)

### Disease Progression & Prognosis

Alzheimer’s disease progresses gradually, with symptoms becoming more severe over time. In the mild or early stage, individuals may experience memory loss, difficulty finding words, trouble managing finances or planning tasks, and may take longer to complete normal daily activities. During the moderate stage, memory loss and confusion become more pronounced, and individuals may have difficulty recognizing family and friends, completing multistep tasks, or adjusting to new situations. Behavioral changes such as agitation, delusions, or impulsivity may also occur. In the severe or late stage, cognitive and physical function decline substantially. Individuals may lose the ability to communicate and become completely dependent on others for help with basic activities of daily living. Alzheimer’s is a progressive and ultimately fatal disease, although the rate of progression and time from diagnosis to death can vary considerably between individuals.

Sources: https://www.nia.nih.gov/health/alzheimers-and-dementia/alzheimers-disease-fact-sheet

### Continuum of Care Providers

Alzheimer’s care typically involves several types of healthcare providers as the disease progresses. A primary care provider (PCP) is often the first point of contact and may perform an initial cognitive screening, review symptoms and medical history, and determine whether further evaluation is needed.

Patients with concerning findings may then be referred to specialists such as a neurologist, geriatrician, or neuropsychologist for more comprehensive diagnostic testing and treatment planning. After diagnosis, ongoing care may involve medication management as well as support from rehabilitation professionals.

Occupational therapists can help patients adapt daily activities and their home environment to maintain independence and safety, while physical therapists may help address mobility, balance, and fall risk. Speech-language pathologists may assist with communication and swallowing problems as the disease progresses. Care needs generally increase over time, often requiring additional involvement from caregivers, social workers, home-care providers, and long-term-care services.

### Biological Mechanisms  

Alzheimer’s involves two key protein abnormalities: the accumulation of amyloid-beta forming plaques between neurons and phosphorylated tau proteins aggregating into tangles inside neurons. Plaque formation can interfere with communication between neurons and tau tangles interfere with the internal transport systems within neurons. As neurons and the connections become increasingly damaged and lost, the brain undergoes atrophy and individuals experience progressive cognitive decline.The formations of plaques and tangles cause synaptic dysfunction and neuroinflammation mediated by microglial activity. Genetic risk factors, such as the APOE ε4 variant is important for late-onset Alzheimer’s and can affect amyloid accumulation, inflammation, and microglial function. Alzheimer’s disease also affects numerous areas of the brain such as the hippocampus, entorhinal cortex, and temporal lobe structures. As the disease progresses, it spreads to the cerebral cortex, frontal and parietal regions, which impacts broader networks involved in language, reasoning, movement, and sensory processing.

Source: https://www.nia.nih.gov/health/alzheimers-causes-and-risk-factors/what-happens-brain-alzheimers-disease

### Clinical Trials / Next-Generation Therapies

Current Alzheimer’s clinical trials are testing treatments that target the disease itself rather than only its symptoms. Anti-amyloid therapies aim to remove amyloid while reducing side effects such as ARIA, while anti-tau therapies target abnormal tau buildup and spread.

Researchers are also studying treatments for neuroinflammation, synaptic dysfunction, and metabolic or mitochondrial changes, as well as vaccines that stimulate an immune response against amyloid or tau. Many newer trials focus on earlier intervention, enrolling people with Alzheimer’s biomarkers before dementia develops, since treatment may be more effective before major neuronal loss has occurred.

## Data Analysis: 
The donor dataset was analyzed using Python to investigate the relationship between years of education and cognitive assessment scores as well as the relationship between Aβ42 levels and dementia status. Pandas was used for data organization, Matplotlib for visualization, and SciPy/scikit-learn for statistical analysis and linear regression.

The first step was to import the dataset (contains demographic, cognitive, genetic, and biomarker information for each donor) into python using the pandas library. 

```python
import csv
import matplotlib.pyplot as plt
from scipy import stats
import numpy as np
import statistics
import pandas as pd
from sklearn.linear_model import LinearRegression

from project_patient import Patient

patients = []

 with open('Metadata and Protein Data for Module 1.csv', 'r') as file: #open the csv file and read the data into a list of patient objects
    reader = csv.DictReader(file)

    for row in reader:
        patient = Patient(
            donor_id=row['Donor ID'],
            age_at_death=float(row['Age at Death']),
            sex=row['Sex'],
            years_of_education=float(row['Years of education']),
            apoe_genotype=row['APOE Genotype'],
            cognitive_status=row['Cognitive Status'],
            thal=row['Thal'],
            abeta40=float(row['ABeta40 pg/ug']),
            abeta42=float(row['ABeta42 pg/ug']),
            ttau=float(row['tTAU pg/ug']),
            ptau=float(row['pTAU pg/ug']),
            mmse_score=float(row['Last MMSE Score']) if row['Last MMSE Score'] != '' else np.nan,
        )

        patients.append(patient)
```

The second step is to clean the data by selecting relevant variables and removing missing values. The relevant variables that were selected included years of education, MMSE scores, cognitive status, and Aβ42 levels. Rows with missing values were removed to ensure that the statistical analyses were performed using complete observations.

```python
# Get years of education and MMSE scores
education = np.array([
    patient.years_of_education
    for patient in patients
    if not np.isnan(patient.mmse_score)
])

#separate abeta42 vlaues into dementia and no dementia groups
dementia_abeta42 = np.array([
    patient.abeta42
    for patient in patients
    if patient.cognitive_status == "Dementia"
])

no_dementia_abeta42 = np.array([
    patient.abeta42
    for patient in patients
    if patient.cognitive_status == "No dementia"
])
```

Then, we separated donors into groups based on cognitive status (dementia and non-dementia groups). This allowed Aβ42 levels to be compared between individuals with and without dementia.
Python code is the same above

The mean Aβ42 concentration was calculated separately for each group so that when comparing these means, it would provide an initial indication of whether Aβ42 levels differ according to dementia status.

dementia_mean = np.mean(dementia_abeta42)

print("Mean Abeta42 for no dementia group: {no_dementia_mean}")

Two visualizations were created to examine patterns and analyze the data. A scatter plot was used to determine the relationship between years of education and MMSE cognitive scores, while a bar graph was used to compare the mean Aβ42 levels between dementia and non-dementia groups.

```python

mmse_scores = np.array([
    patient.mmse_score
    for patient in patients
    if not np.isnan(patient.mmse_score)
])

# Create the linear regression model, reshape education data for the model, and train the model
model = LinearRegression()
education_reshaped = education.reshape(-1, 1)
model.fit(education_reshaped, mmse_scores)

# Predict MMSE scores
predicted_mmse = model.predict(education_reshaped)

# Get slope and intercept
slope = model.coef_[0]
intercept = model.intercept_

print(f"Slope: {slope}")
print(f"Intercept: {intercept}")

# Calculate R-squared
r_squared = model.score(education_reshaped, mmse_scores)

print(f"R-squared: {r_squared}")

# Calculate correlation and p-value
correlation, p_value = stats.pearsonr(education, mmse_scores)

print(f"Correlation: {correlation}")
print(f"P-value: {p_value}")


#independent t-test
t_statistic, p_value_abeta = stats.ttest_ind(
    dementia_abeta42,
    no_dementia_abeta42
)

print(f"T-statistic: {t_statistic}")
print(f"P-value: {p_value_abeta}")
```
