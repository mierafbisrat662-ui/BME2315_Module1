### Prevalence & Incidence

Alzheimer’s disease affects a substantial and growing portion of the older U.S. population. In 2026, an estimated 7.4 million Americans age 65 and older are living with clinical Alzheimer’s dementia, representing about 1 in 9 adults in this age group. The incidence of Alzheimer’s also increases significantly with age. Data from the Chicago Health and Aging Project estimated an annual incidence of approximately 0.4% among adults ages 65–74, rising to 3.2% among adults ages 75–84 and 7.6% among those age 85 and older. These trends demonstrate that age is strongly associated with the likelihood of developing Alzheimer’s disease and help explain why the overall number of Americans affected by the disease continues to increase as the population ages.

### Risk Factors (Genetic, Lifestyle)

Several genetic, biological, and lifestyle factors can increase a person’s risk of developing Alzheimer’s disease. Older age is the greatest known risk factor, with most cases occurring after age 65 and risk increasing as a person gets older. Genetics also play an important role. Variants of the APOE gene can affect Alzheimer’s risk: the APOE ε4 allele increases risk, particularly when a person inherits two copies, while APOE ε2 may provide some protection. Having a family history of the disease is also associated with an increased risk, though neither APOE status nor family history determines whether someone will develop the disease.

Other factors associated with risk include traumatic brain injury, physical inactivity, smoking, an unhealthy diet, and inadequate sleep. Alzheimer’s is also more common among women: almost two-thirds of Americans living with Alzheimer’s are women, although some of this difference is explained by women living longer on average, and researchers continue to study whether biological differences contribute additional risk.

### Symptoms

Alzheimer’s disease causes progressive changes in memory, thinking, judgment, daily functioning, and behavior. Memory loss is one of the most common early symptoms, including forgetting recent events and conversations, repeating statements, misplacing belongings, and becoming lost even in familiar places.

As the disease progresses, people may also have difficulty concentrating and thinking, especially with abstract concepts or numbers, and may struggle with managing finances or completing multiple tasks at the same time. The disease can also impair judgment and decision-making, which can lead to poor choices or difficulty with everyday problems.

Changes in personality and behavior can also occur, including mood changes, depression, social withdrawal, irritability, anger, delusions, and changes in sleeping patterns.

Source: (https://www.mayoclinic.org/diseases-conditions/alzheimers-disease/symptoms-causes/syc-20350447)

### Standard of Care Treatments & Reimbursement

Treatment for Alzheimer’s disease depends on the individual’s symptoms, disease severity, and overall health. Although there is currently no cure, treatment may include medications that temporarily improve memory and thinking, medications that manage behavioral symptoms, and newer therapies that target the underlying biology of the disease and may slow its progression in some patients.

As Alzheimer’s progresses, supportive care becomes increasingly important and may involve caregivers, simplifying daily routines, and modifying the home environment to remove hazards and improve safety.

In terms of reimbursement, Medicare generally covers medically necessary diagnostic testing, physician visits, prescription drugs under applicable plans, and certain skilled nursing or home-health services, but it does not generally pay for long-term custodial residential care. Medicaid may provide broader long-term-care coverage for eligible individuals, including nursing-home care and some home- and community-based services.

Source: (https://www.medicare.gov/basics/costs/help/medicaid)

### Continuum of Care Providers

Alzheimer’s care typically involves several types of healthcare providers as the disease progresses. A primary care provider (PCP) is often the first point of contact and may perform an initial cognitive screening, review symptoms and medical history, and determine whether further evaluation is needed.

Patients with concerning findings may then be referred to specialists such as a neurologist, geriatrician, or neuropsychologist for more comprehensive diagnostic testing and treatment planning. After diagnosis, ongoing care may involve medication management as well as support from rehabilitation professionals.

Occupational therapists can help patients adapt daily activities and their home environment to maintain independence and safety, while physical therapists may help address mobility, balance, and fall risk. Speech-language pathologists may assist with communication and swallowing problems as the disease progresses. Care needs generally increase over time, often requiring additional involvement from caregivers, social workers, home-care providers, and long-term-care services.

### Clinical Trials / Next-Generation Therapies

Current Alzheimer’s clinical trials are testing treatments that target the disease itself rather than only its symptoms. Anti-amyloid therapies aim to remove amyloid while reducing side effects such as ARIA, while anti-tau therapies target abnormal tau buildup and spread.

Researchers are also studying treatments for neuroinflammation, synaptic dysfunction, and metabolic or mitochondrial changes, as well as vaccines that stimulate an immune response against amyloid or tau. Many newer trials focus on earlier intervention, enrolling people with Alzheimer’s biomarkers before dementia develops, since treatment may be more effective before major neuronal loss has occurred.
